Tras realizar la primera aproximación cuantitativa al conjunto de datos de EdificaNova, se validan los siguientes aspectos fundamentales para el negocio y el análisis:

### **2. Limpieza de datos**

El proceso de limpieza e imputación de datos se diseñó para corregir la presencia marginal de valores faltantes (aproximadamente un 4% por columna afectada), asegurando la robustez estadística sin distorsionar la naturaleza del negocio de EdificaNova:

Variables Cualitativas Categóricas (forma_pago y revestimiento): Se utilizó una estrategia de imputación por la moda (el valor más frecuente). Reemplazar los registros vacíos con la categoría dominante previene la pérdida de filas completas en el dataset y mantiene las proporciones reales de la demanda comercial.

Variable Cuantitativa Continua (m2_referencia): Se optó por la imputación mediante la mediana en lugar de la media aritmética. Esta decisión técnica se fundamenta en que la mediana es una métrica insensible a los valores extremos (outliers), evitando sesgar la tendencia central del tamaño de las viviendas solicitadas.

Consistencia en Costos (precio_cotizado): Aquellos registros cuyos precios se encontraban vacíos debido a la falta inicial de metros cuadrados fueron recalculados de manera algorítmica utilizando las reglas de costos de producción de la empresa (considerando los recargos por sistema constructivo, nivel de entrega y distancia logística). Esto garantiza que no existan inconsistencias financieras en el posterior análisis descriptivo.

```python
# ==================================================
#         ETAPA 2: LIMPIEZA Y TRATAMIENTO DE DATOS
# ==================================================

print("=== Conteo Inicial de Valores Faltantes por Columna ===")
print(df.isnull().sum())

# 1. Tratamiento de nulos en variables cualitativas (Imputación por la Moda)
# Reemplazamos los NaN en 'forma_pago' y 'revestimiento' con el valor más frecuente
moda_pago = df['forma_pago'].mode()[0]
moda_revestimiento = df['revestimiento'].mode()[0]

df['forma_pago'].fillna(moda_pago, inplace=True)
df['revestimiento'].fillna(moda_revestimiento, inplace=True)

# 2. Tratamiento de nulos en variables cuantitativas (Imputación por Mediana)
# Para 'm2_referencia', usamos la mediana para evitar que valores extremos afecten el promedio
mediana_m2 = df['m2_referencia'].median()
df['m2_referencia'].fillna(mediana_m2, inplace=True)

# 3. Recalcular precio_cotizado si quedó en nulo por falta de m2
# (Utilizamos la lógica de negocio para no perder la coherencia comercial del registro)
def reparar_precio_nulo(row):
    if pd.isnull(row['precio_cotizado']):
        precio_base = row['m2_referencia'] * 280000
        if row['sistema_constructivo'] == 'S.F': precio_base *= 1.25
        if row['sistema_constructivo'] == 'W.F': precio_base *= 1.15
        if row['tipo_entrega'] == 'LLAVE EN MANO': precio_base *= 1.40
        precio_base += (row['distancia_km'] * 1500)
        return round(precio_base, -3)
    return row['precio_cotizado']

df['precio_cotizado'] = df.apply(reparar_precio_nulo, axis=1)

print("\n=== Conteo Post-Limpieza de Valores Faltantes ===")
print(df.isnull().sum())
```

### **3. Análisis descriptivo**

Interpretación de Hallazgos y Patrones del Negocio
Estratificación del Catálogo de Costos: Al cruzar los sistemas constructivos con los tipos de entrega, se evidencia que el Steel Frame (S.F) bajo la modalidad Llave en Mano representa el producto de mayor valor premium para EdificaNova. Por el contrario, el sistema tradicional E.E en Obra Gris se posiciona como la opción de entrada al mercado, manteniendo una dispersión baja (std), lo que demuestra una alta estandarización en los presupuestos de este segmento.

Validación de Comportamiento Comercial: La tabla de contingencia desmitifica el impacto de la financiación. Aunque los clientes de CONTADO representan un volumen alto de consultas, los leads bajo la modalidad FINANCIACION PROPIA muestran una tasa de conversión a VENDIDO competitiva. Esto sugiere que flexibilizar las opciones de pago no degrada la calidad del lead, sino que expande el universo de cierre comercial.

Análisis del Factor Logístico: Las locaciones fuera de Córdoba Capital (como Buenos Aires o La Plata) muestran las mayores medias en distancia_km. El consecuente incremento en el precio_cotizado demuestra que el algoritmo está absorbiendo correctamente el riesgo de traslado, evitando que la distancia geográfica licúe el margen de ganancia neto de la constructora.

```python
# ==================================================
#    3.1 ANÁLISIS BIVARIADO Y MÉTRICAS DE NEGOCIO
# ==================================================

print("==================================================")
print("   ANÁLISIS DE COSTOS POR CONFIGURACIÓN DE OBRA   ")
print("==================================================")

# 1. ¿Cómo impacta el Sistema Constructivo y el Tipo de Entrega en el precio promedio?
Costo_por_configuracion = df.groupby(['sistema_constructivo', 'tipo_entrega'])['precio_cotizado'].agg(['mean', 'median', 'std']).round(0)
display(Costo_por_configuracion)

print("\n==================================================")
print("      RENDIMIENTO COMERCIAL SEGÚN FORMA DE PAGO    ")
print("==================================================")

# 2. ¿Qué formas de pago tienen mayor tasa de éxito (VENDIDO)?
tabla_contingencia_pago = pd.crosstab(df['forma_pago'], df['estado_lead'], normalize='index') * 100
display(tabla_contingencia_pago.round(2).astype(str) + ' %')

print("\n==================================================")
print("          IMPACTO LOGÍSTICO POR UBICACIÓN          ")
print("==================================================")

# 3. ¿Cuáles son las zonas que nos exigen mayor flete y costo?
logistica_ubicacion = df.groupby('ubicacion')[['distancia_km', 'precio_cotizado']].mean().sort_values(by='distancia_km', ascending=False).round(1)
display(logistica_ubicacion)
```
